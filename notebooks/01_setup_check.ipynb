```python
# Verify counts of images in the 6-class ROI (Region Of Interest) dataset
import os
base_path = "../data/raw/azh/wound_classification-main/data/ROI"
if not os.path.exists(base_path):
    print(f"Directory Path not found: {base_path}")
else:
    for folder in sorted(os.listdir(base_path)):
        folder_path = os.path.join(base_path, folder)
        if os.path.isdir(folder_path):
            num_images = len([f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))])
            print(f"{folder}: {num_images} images")
```
